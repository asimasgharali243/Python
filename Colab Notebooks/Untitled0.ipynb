## Verify Total Spent calculation

### Subtask:
Verify whether Price Per Unit × Quantity = Total Spent. Flag mismatches.

**Reasoning**:
Calculate the expected 'Total Spent' by multiplying 'Price Per Unit' and 'Quantity', then compare it with the actual 'Total Spent'. Flag rows where the calculated and actual values do not match, considering potential floating-point inaccuracies.

```python
# Calculate the expected Total Spent
df['Expected Total Spent'] = df['Price Per Unit'] * df['Quantity']

# Flag mismatches (allowing for a small tolerance due to floating-point arithmetic)
tolerance = 1e-6
mismatched_total_spent = df[
    abs(df['Expected Total Spent'] - df['Total Spent']) > tolerance
]

print(f"Number of rows with mismatched Total Spent: {len(mismatched_total_spent)}")

if not mismatched_total_spent.empty:
    print("\nMismatched Total Spent rows:")
    display(mismatched_total_spent[['Price Per Unit', 'Quantity', 'Total Spent', 'Expected Total Spent']])
else:
    print("\nTotal Spent calculation is consistent for all rows.")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Function to detect outliers using IQR
def detect_outliers_iqr(data):
    Q1 = np.percentile(data, 25)
    print(Q1)
    Q3 = np.percentile(data, 75)
    print(Q3)
    IQR = Q3 - Q1
    print(IQR)
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    print("lower_bound",lower_bound)
    print("upper_bound",upper_bound)
    outliers = [x for x in data if x < lower_bound or x > upper_bound]
    return outliers, lower_bound, upper_bound
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Function to detect outliers using IQR
def detect_outliers_iqr(data):
    Q1 = np.percentile(data, 25)
    print(Q1)
    Q3 = np.percentile(data, 75)
    print(Q3)
    IQR = Q3 - Q1
    print(IQR)
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    print("lower_bound",lower_bound)
    print("upper_bound",upper_bound)
    outliers = [x for x in data if x < lower_bound or x > upper_bound]
    return outliers, lower_bound, upper_bound

# -----------------------------
# Case 1: One Outlier
# Monthly expenses in dollars (one month had a huge hospital bill)
expenses = [2200, 2500, 2300, 2400, 2450, 2550, 2350, 2400, 2380, 2420, 3000, 10000]

outliers1, lb1, ub1 = detect_outliers_iqr(expenses)
print("Case 1: Monthly Expenses")
print("Outliers:", outliers1)

plt.boxplot(expenses)
plt.title("Case 1: Monthly Expenses with One Outlier")
plt.ylabel("Dollars")
plt.show()

```

```python
scores = [50, 55, 60, 65, 70, 100]
plt.boxplot(scores)
plt.title("Boxplot of Scores (with outlier)")
plt.show()
```

```python
import numpy as np

# Calculate IQR
Q1 = np.percentile(scores, 25)
Q3 = np.percentile(scores, 75)
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Remove outliers
scores_cleaned = [x for x in scores if x >= lower_bound and x <= upper_bound]

plt.boxplot(scores_cleaned)
plt.title("Boxplot of Scores (without outlier)")
plt.show()
```

```python
from scipy.stats import zscore

# Calculate the Z-scores for the scores data
z_scores = zscore(scores)

# Define a threshold for identifying outliers (e.g., a Z-score of 2 or 3)
threshold = 2

# Identify outliers
outlier_indices = np.where(np.abs(z_scores) > threshold)[0]
outliers_zscore = [scores[i] for i in outlier_indices]
print(f"Outliers detected using Z-score method (threshold={threshold}): {outliers_zscore}")

# Remove outliers
scores_cleaned_zscore = [scores[i] for i in range(len(scores)) if i not in outlier_indices]

plt.boxplot(scores_cleaned_zscore)
plt.title(f"Boxplot of Scores (without outliers - Z-score threshold={threshold})")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew

# Example data
data_normal = np.random.normal(0, 1, 1000)
data_right = np.random.exponential(1, 1000)
data_left = -np.random.exponential(1, 1000)

# Calculate skewness
print("Normal Skewness:", skew(data_normal))
print("Right Skewness:", skew(data_right))
print("Left Skewness:", skew(data_left))

# Plot
plt.figure(figsize=(12, 4))
plt.subplot(1,3,1)
plt.hist(data_normal, bins=30, color='blue')
plt.title("Normal Distribution")

plt.subplot(1,3,2)
plt.hist(data_right, bins=30, color='green')
plt.title("Positive Skew (Right)")

plt.subplot(1,3,3)
plt.hist(data_left, bins=30, color='red')
plt.title("Negative Skew (Left)")

plt.show()
```

```python
# This script visualizes different types of data skewness using histograms.

import matplotlib.pyplot as plt
import numpy as np

# Set up a figure with three subplots for comparison.
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 6))

# --- 1. Symmetrical Distribution (No Skewness) ---
# Create a dataset with a normal distribution. This is a classic example
# of a symmetrical dataset where the data is evenly distributed around the mean.
data_symmetrical = np.random.normal(loc=0, scale=1, size=10000)

ax1.hist(data_symmetrical, bins=50, color='blue', alpha=0.7)
ax1.set_title('Symmetrical Distribution (No Skewness)')
ax1.set_xlabel('Value')
ax1.set_ylabel('Frequency')
ax1.grid(axis='y', linestyle='--')

# --- 2. Positively Skewed Distribution ---
# Create a dataset with positive skewness. This means the tail of the
# distribution is longer on the right side. We can create this using
# a log-normal distribution, which is naturally positively skewed.
data_positive_skew = np.random.lognormal(mean=0.0, sigma=1.0, size=10000)

ax2.hist(data_positive_skew, bins=50, color='green', alpha=0.7)
ax2.set_title('Positively Skewed Distribution')
ax2.set_xlabel('Value')
ax2.set_ylabel('Frequency')
ax2.grid(axis='y', linestyle='--')

# --- 3. Negatively Skewed Distribution ---
# Create a dataset with negative skewness. This means the tail of the
# distribution is longer on the left side. We can achieve this by inverting
# a positively skewed distribution.
data_negative_skew = -np.random.lognormal(mean=0.0, sigma=1.0, size=10000)

ax3.hist(data_negative_skew, bins=50, color='red', alpha=0.7)
ax3.set_title('Negatively Skewed Distribution')
ax3.set_xlabel('Value')
ax3.set_ylabel('Frequency')
ax3.grid(axis='x', linestyle='--',color='blue')

# Adjust layout to prevent titles from overlapping
plt.tight_layout()

# Display the charts
plt.show()
```

The fnlwgt column represents the "final weight" of the individual in the dataset. It's a sampling weight used by the Census Bureau to make the dataset representative of the U.S. population. While it's important for calculating population-level statistics from this sample, it is generally not meaningful for predicting an individual's income or other characteristics within the dataset itself. Including it directly in a predictive model might not be helpful and could potentially introduce noise or bias, as it's not an inherent characteristic of the individual.
