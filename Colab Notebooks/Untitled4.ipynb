## Understand the basics of linear regression

### Subtask:
Explain the fundamental concepts behind linear regression, including the model equation and the goal of the algorithm.


**Reasoning**:
Provide a markdown explanation covering the fundamental concepts of linear regression as outlined in the instructions.



```python
# Define what linear regression is.
print("Linear regression is a statistical method used to model the relationship between a continuous target variable and one or more independent variables. Its purpose is to predict the value of the target variable based on the values of the independent variables by finding the best-fitting linear relationship.")

# Write out the basic linear regression equation.
print("\nThe basic linear regression equation is:")
print("y = β₀ + β₁x₁ + β₂x₂ + ... + βnxn + ε")
print("\nWhere:")
print("y is the target variable (what you are trying to predict).")
print("β₀ is the y-intercept, the value of y when all independent variables are zero.")
print("β₁, β₂, ..., βn are the coefficients for each independent variable, representing the change in y for a one-unit change in the corresponding independent variable, holding other variables constant.")
print("x₁, x₂, ..., xn are the independent variables (features) used for prediction.")
print("ε is the error term or residual, representing the difference between the actual and predicted values, accounting for random variability and unmeasured factors.")

# Describe the goal of the linear regression algorithm.
print("\nThe goal of the linear regression algorithm is to find the values of the coefficients (β₀, β₁, ..., βn) that minimize the difference between the predicted values (ŷ) and the actual target values (y). This is typically achieved by minimizing the sum of the squared residuals (the difference between actual and predicted values), a method known as Ordinary Least Squares (OLS). This process finds the 'best-fitting' line or hyperplane that represents the linear relationship in the data.")

# Briefly mention the concept of the residual or error term.
print("\nThe residual or error term (ε) in the model equation represents the part of the target variable that cannot be explained by the linear relationship with the independent variables. It is the vertical distance between each data point and the best-fitting line/hyperplane.")
```

## Feature selection

### Subtask:
Explain different methods for selecting relevant features to improve model performance and reduce overfitting.


**Reasoning**:
Explain the importance of feature selection and describe different methods for selecting relevant features for linear regression, including filter, wrapper, and embedded methods.



```python
print("Why Feature Selection is Important for Linear Regression:")
print("Feature selection is the process of choosing a subset of relevant features (independent variables) to use in your model training. For linear regression, it offers several key benefits:")
print("1. Improved Model Performance: Using only the most relevant features can lead to a simpler model that captures the underlying patterns in the data more effectively, potentially resulting in better prediction accuracy.")
print("2. Reduced Overfitting: Including irrelevant or redundant features can cause the model to learn noise in the training data, leading to poor performance on unseen data (overfitting). Feature selection helps to mitigate this by focusing on features with true predictive power.")
print("3. Enhanced Interpretability: A model with fewer features is easier to understand and interpret. This is particularly important in linear regression, where the coefficients represent the impact of each feature on the target variable.")
print("4. Faster Training: Training a model with fewer features is computationally less expensive and faster.")
print("5. Reduced Data Collection/Storage Costs: In some cases, identifying and using only essential features can reduce the cost and effort associated with data collection and storage.")

print("\nCommon Feature Selection Methods for Linear Regression:")
print("There are several approaches to feature selection, broadly categorized as filter, wrapper, and embedded methods:")

print("\nFilter Methods:")
print("Filter methods select features based on their intrinsic properties or their relationship with the target variable, independent of the chosen model. They 'filter' out features before the model training begins. Common techniques include:")
print("- Correlation: Calculating the correlation coefficient between each independent variable and the target variable. Features with high absolute correlation are considered more relevant.")
print("- Statistical Tests: Using statistical tests to assess the relationship between each feature and the target variable. For example, ANOVA (Analysis of Variance) can be used to compare the means of the target variable across different categories of a categorical feature, and t-tests or F-tests can be used for numerical features.")
print("Filter methods are generally computationally less expensive but may not consider the interactions between features.")

print("\nWrapper Methods:")
print("Wrapper methods evaluate subsets of features by training and evaluating a model on each subset. They 'wrap' the feature selection process around the model training. These methods are generally more computationally intensive than filter methods but can capture feature interactions. Common techniques include:")
print("- Forward Selection: Starts with an empty set of features and iteratively adds the feature that provides the most significant improvement in model performance (e.g., based on R-squared or adjusted R-squared) until no further improvement is observed.")
print("- Backward Elimination: Starts with all features and iteratively removes the feature that has the least impact on model performance until no further improvement is observed.")
print("- Recursive Feature Elimination (RFE): Works by recursively removing features and building a model on the remaining features. It uses the model's coefficients (for linear models) or feature importance (for tree-based models) to rank features and eliminate the least important ones.")

print("\nEmbedded Methods:")
print("Embedded methods perform feature selection as part of the model training process itself. These methods have built-in mechanisms to select features. A common example in linear regression is using regularization techniques:")
print("- Lasso (L1 Regularization): Adds a penalty to the linear regression objective function based on the absolute values of the coefficients. This penalty encourages the model to shrink some coefficients to exactly zero, effectively performing feature selection by excluding those features from the model.")
print("- Ridge (L2 Regularization): Adds a penalty based on the squared values of the coefficients. While Ridge shrinks coefficients, it rarely shrinks them to exactly zero, so it primarily helps with multicollinearity rather than strict feature selection.")
print("Embedded methods are often less computationally expensive than wrapper methods and can capture some feature interactions.")

print("\nConsiderations When Choosing a Feature Selection Method:")
print("The choice of feature selection method depends on several factors:")
print("- Computational Cost: Filter methods are generally the fastest, followed by embedded methods, and then wrapper methods.")
print("- Dataset Size: For very large datasets, filter methods might be more practical due to their lower computational cost. Wrapper methods can become computationally prohibitive with a large number of features.")
print("- Need for Interpretability: Filter methods based on simple metrics like correlation can be highly interpretable. Wrapper and embedded methods might be less directly interpretable in terms of why certain features were selected.")
print("- Feature Interactions: Wrapper and embedded methods are generally better at capturing interactions between features compared to filter methods.")
```

## Train the linear regression model

### Subtask:
Detail the process of training the model using the prepared data, including the optimization algorithm used to find the best model parameters.


**Reasoning**:
Detail the process of training a linear regression model, including the role of the loss function and the OLS optimization algorithm.



```python
print("Training a Linear Regression Model:")
print("Once the data is prepared and features are selected, the next step is to train the linear regression model using the training data. The goal of training is to find the optimal values for the model's parameters (the coefficients β₁, β₂, ..., βn and the intercept β₀) that best fit the training data.")

print("\nThe Goal: Minimizing the Difference Between Predicted and Actual Values:")
print("The core idea is to find the line (or hyperplane in higher dimensions) that minimizes the overall difference between the values predicted by the model (ŷ) and the actual target values (y) in the training dataset. For a given data point, the difference between the actual value (yᵢ) and the predicted value (ŷᵢ) is called the residual (eᵢ = yᵢ - ŷᵢ).")

print("\nThe Role of the Loss Function (Mean Squared Error):")
print("To quantify this overall difference across all data points in the training set, we use a loss function (also known as a cost function or objective function). The loss function measures how well the model is performing. For linear regression, the most commonly used loss function is the Mean Squared Error (MSE).")
print("\nMSE is calculated as the average of the squared residuals:")
print("MSE = (1/n) * Σ(yᵢ - ŷᵢ)²")
print("\nWhere:")
print("n is the number of data points in the training set.")
print("yᵢ is the actual target value for the i-th data point.")
print("ŷᵢ is the predicted target value for the i-th data point.")
print("Σ denotes the sum over all data points.")
print("\nWe square the residuals for two main reasons:")
print("1. To ensure that positive and negative errors do not cancel each other out.")
print("2. To penalize larger errors more heavily, as squaring amplifies larger differences.")
print("\nThe training process for linear regression is essentially an optimization problem: find the values of β₀, β₁, ..., βn that minimize the MSE (or equivalently, the Sum of Squared Residuals - SSR = Σ(yᵢ - ŷᵢ)²).")

print("\nOptimization Algorithm: Ordinary Least Squares (OLS):")
print("For simple linear regression and even multiple linear regression, when the assumptions of OLS are met, there is a closed-form solution to find the parameters that minimize the Sum of Squared Residuals (SSR). This method is called Ordinary Least Squares (OLS).")
print("\nHow OLS Works (Conceptually):")
print("OLS finds the unique line (or hyperplane) that minimizes the sum of the vertical distances (squared) from each data point to the line. Imagine drawing a line through your data points; OLS finds the line where the sum of the areas of the squares built on the vertical distances from each point to the line is as small as possible.")
print("\nHow OLS Works (Mathematically - Brief Mention):")
print("Mathematically, OLS finds the parameters (β₀, β₁, ..., βn) by solving a set of linear equations derived from setting the partial derivatives of the SSR with respect to each parameter to zero. This results in a direct, analytical solution for the optimal parameter values. In matrix form, the solution for the coefficients (β) is given by: β = (XᵀX)⁻¹Xᵀy, where X is the matrix of independent variables (with a column of ones for the intercept), Xᵀ is its transpose, (XᵀX)⁻¹ is the inverse of XᵀX, and y is the vector of target values.")

print("\nLearning the Coefficients and Intercept:")
print("The OLS algorithm uses the training data (the values of the independent variables X and the actual target values y) to calculate the optimal values for the coefficients (β₁, β₂, ..., βn) and the intercept (β₀) using the closed-form solution. These calculated parameters represent the best-fitting line/hyperplane for the training data in terms of minimizing the squared errors.")

print("\nOther Potential Optimization Methods (Brief Mention):")
print("While OLS provides a direct solution for standard linear regression, for very large datasets or more complex linear models (like those with regularization where a closed-form solution might not exist or be computationally expensive to calculate directly), iterative optimization algorithms like Gradient Descent can be used.")
print("- Gradient Descent: An iterative algorithm that starts with initial parameter values and repeatedly adjusts them in the direction that decreases the loss function (MSE) until a minimum is reached. It calculates the gradient (the direction of steepest ascent) of the loss function with respect to the parameters and moves in the opposite direction (steepest descent).")
```

## Evaluate the model

### Subtask:
Describe common metrics for evaluating the performance of a linear regression model, such as Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R-squared.


**Reasoning**:
Explain the purpose of model evaluation and describe MSE, RMSE, and R-squared as evaluation metrics for linear regression, including their formulas, interpretations, and relationship to each other.



```python
print("Evaluating the Performance of a Linear Regression Model:")
print("After training a linear regression model, it is crucial to evaluate its performance to understand how well it can predict the target variable on unseen data. Model evaluation provides insights into the model's accuracy, reliability, and generalization ability. It helps in comparing different models, tuning hyperparameters, and ensuring the model is suitable for the intended purpose.")

print("\nCommon Evaluation Metrics for Linear Regression:")
print("Several metrics are used to evaluate the performance of linear regression models. These metrics quantify the difference between the actual target values (y) and the values predicted by the model (ŷ) on the test dataset.")

print("\n1. Mean Squared Error (MSE):")
print("MSE is one of the most common metrics and measures the average of the squared differences between the actual and predicted values.")
print("Formula:")
print("MSE = (1/n) * Σ(yᵢ - ŷᵢ)²")
print("\nWhere:")
print("n is the number of data points in the dataset.")
print("yᵢ is the actual target value for the i-th data point.")
print("ŷᵢ is the predicted target value for the i-th data point.")
print("Σ denotes the sum over all data points.")
print("\nInterpretation:")
print("A lower MSE indicates a better fit of the model to the data. MSE is always non-negative. The squaring of errors makes larger errors more prominent, penalizing outliers more heavily.")
print("The unit of MSE is the square of the unit of the target variable, which can make it less intuitive to interpret in the context of the original data scale.")

print("\n2. Root Mean Squared Error (RMSE):")
print("RMSE is the square root of the MSE. It is also a widely used metric.")
print("Formula:")
print("RMSE = √MSE = √[(1/n) * Σ(yᵢ - ŷᵢ)²]")
print("\nInterpretation:")
print("RMSE has the same unit as the target variable, making it easier to interpret than MSE. A lower RMSE indicates a better model performance. RMSE represents the typical distance between the actual and predicted values.")
print("RMSE is often preferred over MSE because its value is in the same scale as the target variable, providing a more direct understanding of the magnitude of the prediction errors.")

print("\n3. R-squared (Coefficient of Determination):")
print("R-squared is a metric that represents the proportion of the variance in the dependent variable that is predictable from the independent variables. It indicates how well the independent variables explain the variation in the target variable.")
print("Formula:")
print("R² = 1 - (SSR / SST)")
print("\nWhere:")
print("SSR (Sum of Squared Residuals) = Σ(yᵢ - ŷᵢ)²  (This is the sum of the squared differences between actual and predicted values, also known as the unexplained variance)")
print("SST (Total Sum of Squares) = Σ(yᵢ - ȳ)²  (This is the sum of the squared differences between the actual values and the mean of the actual values, representing the total variance in the target variable)")
print("ȳ is the mean of the actual target values.")
print("\nInterpretation:")
print("R-squared values range from 0 to 1.")
print("An R-squared of 0 means that the model explains none of the variability in the target variable (the model is no better than predicting the mean of the target variable).")
print("An R-squared of 1 means that the model explains all of the variability in the target variable (a perfect fit).")
print("A higher R-squared generally indicates a better fit, meaning the independent variables are good predictors of the target variable. However, a high R-squared doesn't necessarily mean the model is good or the causal relationships are valid; it should be interpreted in the context of the specific problem.")
print("R-squared can sometimes decrease if you add features that do not help the model.")

print("\nOther Potential Evaluation Metrics:")
print("- Mean Absolute Error (MAE): MAE is the average of the absolute differences between the actual and predicted values. It is less sensitive to outliers than MSE and RMSE.")
print("  Formula: MAE = (1/n) * Σ|yᵢ - ŷᵢ|")
print("MAE is in the same unit as the target variable and provides a straightforward average magnitude of errors.")
```

## Tune the model

### Subtask:
Explain techniques for tuning the model, although linear regression has limited hyperparameters, discuss concepts like regularization (Lasso and Ridge) as a form of tuning.


**Reasoning**:
The current subtask is to explain tuning techniques for linear regression, focusing on regularization. I will write a detailed explanation covering the limited tuning options for standard linear regression, introduce Lasso and Ridge regularization as tuning methods to address overfitting and multicollinearity, describe each method with its penalty term and effect on coefficients (including Lasso's feature selection capability), and explain the role of the regularization parameter (alpha).



```python
print("Tuning a Linear Regression Model:")
print("Standard linear regression, as described with the Ordinary Least Squares (OLS) method, has very few hyperparameters to 'tune' in the traditional sense.")
print("Once the features are selected, the OLS method directly calculates the unique set of coefficients that minimize the Mean Squared Error (MSE) on the training data. There are no parameters within the OLS algorithm itself that you typically adjust to change the model's complexity or behavior, other than the choice of features.")
print("The primary ways to influence a standard linear regression model's performance are through data preparation (handling missing values, encoding), feature engineering, and feature selection.")

print("\nRegularization as a Form of Tuning:")
print("While OLS linear regression itself has limited tuning parameters, regularization techniques can be applied to linear regression models. These techniques introduce additional terms to the loss function (the function being minimized during training) that penalize large coefficients.")
print("Regularization helps to 'tune' the model by controlling its complexity and preventing overfitting, especially when dealing with datasets that have many features or significant multicollinearity (high correlation between independent variables). It can also indirectly act as a form of feature selection.")

print("\nLasso Regularization (L1 Regularization):")
print("Lasso (Least Absolute Shrinkage and Selection Operator) is a regularization technique that adds a penalty term based on the absolute values of the coefficients to the standard OLS loss function.")
print("The objective function for Lasso regression is:")
print("Minimize: MSE + α * Σ|βᵢ|")
print("\nWhere:")
print("MSE is the Mean Squared Error (or Sum of Squared Residuals).")
print("α (alpha) is the regularization parameter (α ≥ 0).")
print("Σ|βᵢ| is the sum of the absolute values of the coefficients (excluding the intercept β₀).")
print("\nEffect on Coefficients:")
print("The L1 penalty encourages the model to shrink some coefficients towards zero. Crucially, Lasso can shrink coefficients *exactly* to zero. This effectively removes the corresponding features from the model, performing automatic feature selection. This results in a sparser model, which can be beneficial for interpretability and efficiency.")
print("Lasso is particularly useful when you suspect that many features are irrelevant or redundant.")

print("\nRidge Regularization (L2 Regularization):")
print("Ridge regression is another regularization technique that adds a penalty term based on the squared values of the coefficients to the standard OLS loss function.")
print("The objective function for Ridge regression is:")
print("Minimize: MSE + α * Σ(βᵢ)²")
print("\nWhere:")
print("MSE is the Mean Squared Error (or Sum of Squared Residuals).")
print("α (alpha) is the regularization parameter (α ≥ 0).")
print("Σ(βᵢ)² is the sum of the squared values of the coefficients (excluding the intercept β₀).")
print("\nEffect on Coefficients:")
print("The L2 penalty encourages the model to shrink coefficients towards zero. Unlike Lasso, Ridge rarely shrinks coefficients *exactly* to zero; it shrinks them asymptotically closer to zero. This means that Ridge tends to keep all features in the model but reduces their impact.")
print("Ridge is particularly useful for dealing with multicollinearity. Highly correlated features will share the responsibility for explaining the variance in the target variable, and their coefficients will be shrunk together, making the model more robust to small changes in the training data.")

print("\nThe Regularization Parameter (α):")
print("The parameter α (alpha) is the key tuning parameter for both Lasso and Ridge regression. It controls the strength of the regularization penalty.")
print("- If α = 0: The penalty term is zero, and the model is equivalent to standard OLS linear regression.")
print("- If α > 0: A penalty is applied. Increasing the value of α increases the strength of the penalty.")
print("\nTuning α:")
print("Tuning the α parameter involves finding the optimal value that balances the trade-off between fitting the training data well and keeping the coefficients small (reducing complexity and preventing overfitting).")
print("- A small α allows the model to fit the training data more closely but increases the risk of overfitting.")
print("- A large α heavily penalizes large coefficients, leading to a simpler model with smaller coefficients, which reduces the risk of overfitting but might underfit the training data.")
print("The optimal value of α is typically found using techniques like cross-validation, where the model is trained with different α values on different subsets of the training data, and the performance is evaluated on a validation set.")
```

## Interpret the model

### Subtask:

**Reasoning**:
Explain the meaning and interpretation of linear regression coefficients and the intercept, including caveats.



```python
print("Interpreting the Model's Coefficients:")
print("In a linear regression model, the coefficients represent the estimated change in the target variable for a one-unit increase in the corresponding independent variable, assuming all other independent variables are held constant. This 'holding other variables constant' is crucial for correct interpretation, as it isolates the effect of a single feature.")

print("\nInterpreting the Intercept (β₀):")
print("The intercept term (β₀) represents the expected value of the target variable when all independent variables (x₁, x₂, ..., xn) are zero. Its interpretation is meaningful only if it is plausible for all independent variables to be zero simultaneously and if the training data includes values close to zero for all features. In many real-world scenarios, interpreting the intercept directly might not be practically relevant if the point where all features are zero is outside the range of the observed data.")

print("\nImplications of the Sign and Magnitude of Coefficients:")
print("- Sign: The sign of a coefficient (+ or -) indicates the direction of the relationship between the independent variable and the target variable. A positive coefficient means that as the independent variable increases, the target variable is expected to increase. A negative coefficient means that as the independent variable increases, the target variable is expected to decrease.")
print("- Magnitude: The magnitude (absolute value) of a coefficient indicates the strength of the relationship. A larger magnitude suggests that a one-unit change in the independent variable has a larger estimated impact on the target variable, assuming the variables are on comparable scales. However, comparing the magnitudes of coefficients for features on vastly different scales can be misleading.")

print("\nThe Concept of 'Holding Other Variables Constant':")
print("When interpreting a coefficient βᵢ for an independent variable xᵢ, the interpretation 'a one-unit increase in xᵢ is associated with a βᵢ change in the target variable' is valid *only* when the values of all other independent variables (x₁, ..., xᵢ₋₁, xᵢ₊₁, ..., xn) remain unchanged. This is because linear regression models the *linear combination* of the effects of all features. The coefficient reflects the marginal effect of a single feature, controlling for the influence of other features in the model.")

print("\nCaveats in Interpreting Coefficients:")
print("While coefficients provide valuable insights into the relationships within the model, it's important to be aware of potential pitfalls and limitations:")
print("1. Assumption of Linearity: Linear regression assumes a linear relationship between the independent variables and the target variable. If the true relationship is non-linear, the linear coefficients may not accurately represent the relationship.")
print("2. Multicollinearity: Multicollinearity occurs when independent variables are highly correlated with each other. High multicollinearity can make the coefficients unstable and difficult to interpret. Small changes in the data can lead to large changes in the coefficient estimates. While the model might still have good predictive power, the individual coefficient interpretations become unreliable.")
print("3. Correlation Does Not Imply Causation: A significant coefficient indicates a statistical association (correlation) between the independent variable and the target variable *within the context of the model*. It does *not* imply that the independent variable *causes* the change in the target variable. Establishing causation requires careful experimental design or domain knowledge, not just statistical association from an observational study.")
print("4. Outliers and Influential Points: Outliers or influential data points can disproportionately affect the estimated coefficients, potentially leading to misleading interpretations.")
print("5. Scale of Features: As mentioned, comparing the magnitudes of coefficients for features on different scales is not directly informative about their relative importance. Feature scaling (standardization or normalization) can help make coefficient magnitudes more comparable, but interpretation should still focus on the scaled units.")
print("6. Model Misspecification: If important variables are omitted from the model or if the functional form is incorrect, the estimated coefficients for the included variables may be biased and their interpretations inaccurate.")
```

## Summary:

### Data Analysis Key Findings

*   The basic linear regression equation is \$y = \beta\_0 + \beta\_1x\_1 + ... + \beta\_nx\_n + \epsilon\$, where the goal is to find coefficients (\$beta\$) that minimize the difference between predicted and actual values, typically using Ordinary Least Squares (OLS).
*   Data preparation for linear regression involves handling missing values (imputation or removal), encoding categorical variables (one-hot or label encoding), and splitting data into training and testing sets (commonly 70/30 or 80/20) for unbiased evaluation and detecting overfitting. Other steps like feature scaling and engineering may also be necessary.
*   Feature selection improves linear regression models by enhancing performance, reducing overfitting, and improving interpretability. Methods include Filter (e.g., correlation), Wrapper (e.g., forward selection, RFE), and Embedded methods (e.g., Lasso regularization).
*   Training involves minimizing a loss function, usually Mean Squared Error (MSE), which is the average of squared residuals. OLS provides a closed-form solution for standard linear regression parameters by minimizing MSE. Iterative methods like Gradient Descent can be used for larger datasets or regularized models.
*   Model performance is evaluated using metrics like Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R-squared (\$R^2\$). MSE and RMSE quantify the average magnitude of prediction errors (lower is better), while \$R^2\$ indicates the proportion of target variable variance explained by the model (higher is generally better, from 0 to 1). Mean Absolute Error (MAE) is another metric less sensitive to outliers.
*   Standard OLS linear regression has limited tuning parameters beyond feature selection. Regularization techniques like Lasso (L1) and Ridge (L2) are used as tuning methods. Lasso shrinks coefficients to zero for feature selection, while Ridge shrinks them towards zero to handle multicollinearity. The regularization strength is controlled by the parameter \$\alpha\$, typically tuned using cross-validation.
*   Interpreting coefficients involves understanding that \$\beta\_i\$ represents the estimated change in the target variable for a one-unit increase in \$x\_i\$, holding other features constant. The intercept (\$beta\_0\$) is the expected target value when all features are zero. The sign indicates the relationship direction, and magnitude indicates strength (on comparable scales). Interpretation requires caution due to assumptions like linearity, potential multicollinearity, the distinction between correlation and causation, and the influence of outliers.

### Insights or Next Steps

*   Understanding the assumptions of linear regression (linearity, independence of errors, homoscedasticity, normality of errors, no perfect multicollinearity) is crucial before interpreting coefficients or trusting model performance.
*   For practical application, evaluating the model on the test set using metrics like RMSE and R-squared provides a realistic estimate of how the model will perform on new, unseen data.

```python
from sklearn.linear_model import Lasso, Ridge
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_regression
import numpy as np
import pandas as pd

# Generate a synthetic dataset for demonstration
X_synth, y_synth = make_regression(n_samples=100, n_features=20, n_informative=10, random_state=42)
X_synth = pd.DataFrame(X_synth, columns=[f'Feature_{i}' for i in range(X_synth.shape[1])])
y_synth = pd.Series(y_synth, name='Target')

print("Shape of synthetic data (X_synth, y_synth):", X_synth.shape, y_synth.shape)

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X_synth, y_synth, test_size=0.2, random_state=42
)

# Scale the features - important for regularization
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("\n--- Implementing Regularization (Tuning) ---")

# --- Lasso Regression (L1 Regularization) ---
print("\nLasso Regression:")

# Initialize the Lasso model with a chosen alpha value
# alpha = 0.1 is just an example; this would typically be tuned using cross-validation
lasso_model = Lasso(alpha=0.1, random_state=42)

# Train the Lasso model
lasso_model.fit(X_train_scaled, y_train)

print("Lasso Coefficients:")
# Display non-zero coefficients for better readability
lasso_coefficients = pd.Series(lasso_model.coef_, index=X_train.columns)
display(lasso_coefficients[lasso_coefficients != 0])
print(f"Number of non-zero coefficients (features selected by Lasso): {len(lasso_coefficients[lasso_coefficients != 0])}")

# Evaluate Lasso model (e.g., using cross-validation on the training set)
# Cross-validation helps estimate model performance and tune alpha
# We'll use negative MSE because cross_val_score by default maximizes the score
lasso_scores = cross_val_score(lasso_model, X_train_scaled, y_train, cv=5, scoring='neg_mean_squared_error')
print(f"Mean Cross-Validation MSE for Lasso (alpha=0.1): {-np.mean(lasso_scores):.2f}")


# --- Ridge Regression (L2 Regularization) ---
print("\nRidge Regression:")

# Initialize the Ridge model with a chosen alpha value
# alpha = 1.0 is just an example; this would typically be tuned using cross-validation
ridge_model = Ridge(alpha=1.0, random_state=42)

# Train the Ridge model
ridge_model.fit(X_train_scaled, y_train)

print("Ridge Coefficients (all will likely be non-zero but shrunk):")
ridge_coefficients = pd.Series(ridge_model.coef_, index=X_train.columns)
display(ridge_coefficients)

# Evaluate Ridge model using cross-validation
ridge_scores = cross_val_score(ridge_model, X_train_scaled, y_train, cv=5, scoring='neg_mean_squared_error')
print(f"Mean Cross-Validation MSE for Ridge (alpha=1.0): {-np.mean(ridge_scores):.2f}")

# Note: In a real scenario, you would use techniques like GridSearchCV or RandomizedSearchCV
# with cross-validation to find the optimal alpha for both Lasso and Ridge.
```

```python
import pandas as pd
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.linear_model import Lasso, Ridge
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_regression

# Generate a synthetic dataset for demonstration
# This dataset will have some informative and some uninformative features
X_synth, y_synth = make_regression(n_samples=100, n_features=20, n_informative=5, random_state=42)
X_synth = pd.DataFrame(X_synth, columns=[f'Feature_{i}' for i in range(X_synth.shape[1])])
y_synth = pd.Series(y_synth, name='Target')

print("Shape of synthetic data (X_synth, y_synth):", X_synth.shape, y_synth.shape)
display(X_synth.head())
display(y_synth.head())

# Split the synthetic data for demonstration
X_train_synth, X_test_synth, y_train_synth, y_test_synth = train_test_split(
    X_synth, y_synth, test_size=0.2, random_state=42
)

print("\n--- Feature Selection Methods ---")

# --- Filter Method: SelectKBest with f_regression ---
# Select the top k features based on a statistical test (F-test for regression)
k = 5 # Number of features to select

selector_filter = SelectKBest(score_func=f_regression, k=k)
X_train_filter = selector_filter.fit_transform(X_train_synth, y_train_synth)
X_test_filter = selector_filter.transform(X_test_synth)

print(f"\nFilter Method (SelectKBest with f_regression) - Selected {k} features:")
# Get the indices of the selected features
selected_feature_indices_filter = selector_filter.get_support(indices=True)
selected_feature_names_filter = X_train_synth.columns[selected_feature_indices_filter]
print("Selected features:", selected_feature_names_filter.tolist())
print("Shape after filtering (X_train_filter, X_test_filter):", X_train_filter.shape, X_test_filter.shape)


# --- Embedded Method: Lasso Regularization ---
# Lasso performs feature selection by shrinking coefficients to zero
# We'll use a Lasso model and examine its coefficients

# It's often beneficial to scale features before applying regularization
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_synth)
X_test_scaled = scaler.transform(X_test_synth)

# Train a Lasso model
# The 'alpha' parameter controls the strength of the penalty.
# A higher alpha leads to more coefficients being shrunk to zero.
# We'll use a relatively small alpha here for demonstration purposes.
lasso_model = Lasso(alpha=0.1, random_state=42)
lasso_model.fit(X_train_scaled, y_train_synth)

print("\nEmbedded Method (Lasso Regularization) - Coefficients:")
# Create a Series to easily view coefficients alongside feature names
lasso_coefficients = pd.Series(lasso_model.coef_, index=X_train_synth.columns)
print(lasso_coefficients)

# Features with coefficients equal to zero have been effectively removed by Lasso
selected_features_lasso = lasso_coefficients[lasso_coefficients != 0].index.tolist()
print(f"\nFeatures selected by Lasso (non-zero coefficients): {selected_features_lasso}")
print(f"Number of features selected by Lasso: {len(selected_features_lasso)}")

# Note: Wrapper methods like RFE are more computationally intensive and
# might be less suitable for a simple inline example like this, but they
# involve training the model iteratively with different feature subsets.
```
