## Load and preprocess the data

### Subtask:
Load the images and their corresponding celebrity labels from the "vishesh1412/celebrity-face-image-dataset" dataset. Preprocess the images, including resizing and normalization, to prepare them for model training.


**Reasoning**:
Load the dataset and preprocess the images by resizing and normalizing them, then store the images and labels in lists.



```python
from datasets import load_dataset
from PIL import Image
import numpy as np

# Load the dataset
dataset = load_dataset("vishesh1412/celebrity-face-image-dataset")

# Define the target image size
image_size = (128, 128)

# Lists to store preprocessed images and labels
preprocessed_images = []
labels = []

# Iterate through the dataset and preprocess images
for item in dataset['train']:
    image = item['image']
    label = item['label']

    # Resize the image
    image = image.resize(image_size)

    # Convert image to numpy array and normalize pixel values
    image_array = np.array(image).astype(np.float32) / 255.0

    preprocessed_images.append(image_array)
    labels.append(label)

# Convert lists to numpy arrays
preprocessed_images = np.array(preprocessed_images)
labels = np.array(labels)

print("Preprocessing complete. Shape of preprocessed images:", preprocessed_images.shape)
print("Shape of labels:", labels.shape)
```
