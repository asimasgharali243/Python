```python
consume_unit = int(input("Enter a consume unit: "))

price_per_unit = {
    "f100": 30,   # per unit for first 100
    "f200": 40,   # per unit for next 100 (101–200)
    "f300": 60,   # per unit for next 100 (201–300)
    "f400": 80    # per unit above 300
}

tax = 200
total_price = 0
```

```python
if consume_unit <= 100:
    total_price = consume_unit * price_per_unit["f100"]

elif consume_unit <= 200:
    total_price = (100 * price_per_unit["f100"]) + ((consume_unit % 100) * price_per_unit["f200"])

elif consume_unit <= 300:
    total_price = (100 * price_per_unit["f100"]) + (100 * price_per_unit["f200"]) + ((consume_unit % 200) * price_per_unit["f300"])

else:
    total_price = (100 * price_per_unit["f100"])  + (100 * price_per_unit["f200"])  + (100 * price_per_unit["f300"]) + ((consume_unit % 300) * price_per_unit["f400"])

print("Total Bill:", total_price)
if total_price > 10000: 
    tax = total_price * 15 / 100
total_price=total_price+tax
print(total_price)
```
