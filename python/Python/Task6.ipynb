```python


class BankAccount:
    def __init__(self, account_number, balance, pin):
        self.account_number = account_number   # Public
        self._balance = balance                # Protected
        self.__pin = pin                       # Private

    # ---------- Getter ----------
    @property
    def balance(self):
        """Getter for balance"""
        return self._balance

    # ---------- Setter ----------
    @balance.setter
    def balance(self, amount):
        """Setter with validation"""
        if amount < 0:
            raise ValueError(" Balance cannot be negative")
        self._balance = amount

    # ---------- Public Methods ----------
    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            print(f"Deposited {amount}, new balance = {self._balance}")
        else:
            print("Deposit must be positive")

    def withdraw(self, amount, pin):
        if pin != self.__pin:
            print(" Invalid PIN")
            return
        if amount <= 0:
            print(" Withdrawal must be positive")
            return
        if amount > self._balance:
            print(" Insufficient funds")
            return
        self._balance -= amount
        print(f" Withdrawn {amount}, remaining balance = {self._balance}")

    def change_pin(self, old_pin, new_pin):
        if old_pin == self.__pin:
            self.__pin = new_pin
            print(" PIN changed successfully")
        else:
            print(" Incorrect old PIN")


#  Interactive Menu
def main():
    acountNo=input("enter a account number ")
    DepositeAmount=int(input("enter  a amount"))
    Pin=input("enter a pin")
    acc = BankAccount(acountNo, DepositeAmount, Pin)

    while True:
        print("\n--- Bank Menu ---")
        print("1. Deposit")
        print("2. Withdraw")
        print("3. Check Balance")
        print("4. Change PIN")
        print("5. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            amt = float(input("Enter deposit amount: "))
            acc.deposit(amt)

        elif choice == "2":
            amt = float(input("Enter withdrawal amount: "))
            pin = input("Enter PIN: ")
            acc.withdraw(amt, pin)

        elif choice == "3":
            print(" Current Balance:", acc.balance)

        elif choice == "4":
            old_pin = input("Enter old PIN: ")
            new_pin = input("Enter new PIN: ")
            acc.change_pin(old_pin, new_pin)

        elif choice == "5":
            print(" Exiting. Thank you for using our bank!")
            break

        else:
            print("Invalid choice, try again!")


# Run Program
if __name__ == "__main__":
    main()
```

```python
class Student:
    def __init__(self, name, rollNo, address):
        self.name = name
        self.rollNo = rollNo
        self.address = address
        self.marks = {}
        self.grade = ""

    # ---------- Setters ----------
    def setName(self, name):
        self.name = name

    def setRollNo(self, rollNo):
        self.rollNo = rollNo

    def setAddress(self, address):
        self.address = address

    def setMarks(self, marks):
        self.marks = marks

    # ---------- Getters ----------
    def getName(self):
        return self.name

    def getRollNo(self):
        return self.rollNo

    def getAddress(self):
        return self.address

    def getMarks(self):
        return self.marks

    # ---------- Other Methods ----------
    def add_marks(self, subject, score):
        """Add or update marks for a subject"""
        self.marks[subject] = score

    def calculate_grade(self):
        """Calculate grade based on average marks"""
        if not self.marks:
            self.grade = "No Marks"
            return self.grade

        total = sum(self.marks.values())
        avg = total / len(self.marks)

        if avg >= 90:
            self.grade = "A"
        elif avg >= 75:
            self.grade = "B"
        elif avg >= 60:
            self.grade = "C"
        elif avg >= 40:
            self.grade = "D"
        else:
            self.grade = "F"
        return self.grade

    def display(self):
        """Display student details"""
        print("\n--- Student Details ---")
        print(f"Name    : {self.name}")
        print(f"Roll No : {self.rollNo}")
        print(f"Address : {self.address}")
        print(f"Marks   : {self.marks}")
        print(f"Grade   : {self.grade}")


# ---------- Derived Class ----------
class GraduateStudent(Student):
    def __init__(self, name, rollNo, address, thesis_title):
        super().__init__(name, rollNo, address)
        self.thesis_title = thesis_title

    def setThesisTitle(self, thesis_title):
        self.thesis_title = thesis_title

    def getThesisTitle(self):
        return self.thesis_title

    def display(self):
        """Extended display"""
        super().display()
        print(f"Thesis  : {self.thesis_title}")


# ---------- Menu Driven Program ----------
def main():
    print(" Enter Student Information")
    name = input("Enter name: ")
    rollNo = input("Enter roll number: ")
    address = input("Enter address: ")
    thesis = input("Enter thesis title (for graduate student): ")

    student = GraduateStudent(name, rollNo, address, thesis)

    while True:
        print("\n--- Menu ---")
        print("1. Add/Update Marks")
        print("2. Calculate Grade")
        print("3. Display Student Info")
        print("4. Change Name")
        print("5. Change Address")
        print("6. Change Thesis Title")
        print("7. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            subject = input("Enter subject: ")
            score = int(input("Enter marks: "))
            student.add_marks(subject, score)

        elif choice == "2":
            grade = student.calculate_grade()
            print(f" Grade calculated: {grade}")

        elif choice == "3":
            student.display()

        elif choice == "4":
            new_name = input("Enter new name: ")
            student.setName(new_name)

        elif choice == "5":
            new_address = input("Enter new address: ")
            student.setAddress(new_address)

        elif choice == "6":
            new_thesis = input("Enter new thesis title: ")
            student.setThesisTitle(new_thesis)

        elif choice == "7":
            print(" Exiting...")
            break

        else:
            print(" Invalid choice, try again!")


if __name__ == "__main__":
    main()
```

```python
class Employer:
    def __init__(self, empId, name, salary, phone, address,status):
        self.empId = empId
        self.name = name
        self._salary = salary
        self.phone = phone
        self.address = address
        self.status =status

    # Setters
    def setEmpId(self, empId):
        self.empId = empId

    def setName(self, name):
        self.name = name

    def setSalary(self, salary):
        self._salary = salary

    def setPhone(self, phone):
        self.phone = phone

    def setAddress(self, address):
        self.address = address    
    def setStatus(status):
        self.status=status

    # Getters
    def getEmpId(self):
        return self.empId

    def getName(self):
        return self.name

    def getSalary(self):
        return self._salary

    def getPhone(self):
        return self.phone

    def getAddress(self):
        return self.address    
    def getStatus(self):
        return self.status
    # Display
    def display(self):
        print("\n--- Employer Details ---")
        print(f"ID      : {self.empId}")
        print(f"Name    : {self.name}")
        print(f"Salary  : {self._salary}")
        print(f"Phone   : {self.phone}")
        print(f"Address : {self.address}")
        print(f"Status  :  {self.status}")

class Manager(Employer):
    def __init__(self, empId, name, salary, phone, address,status, department):
        super().__init__(empId, name, salary, phone, address,status)
        self.department = department

    def display(self):
        super().display()
        print(f"Dept    : {self.department}")


class Developer(Employer):
    def __init__(self, empId, name, salary, phone, address,status, programmingLanguage):
        super().__init__(empId, name, salary, phone, address,status)
        self.programmingLanguage = programmingLanguage
          
    def display(self):
        super().display()
        print(f"Programming Language : {self.programmingLanguage}")


def main():
    employees = []

    while True:
        print("\n--- Employee Menu ---")
        print("1. Add Manager")
        print("2. Add Developer")
        print("3. Update Salary")
        print("4. Display All Employees")
        print("5. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            name = input("Enter manager name: ")
            emp_id = input("Enter ID: ")
            salary = float(input("Enter salary: "))
            dept = input("Enter department: ")
            phone=input("Enter a Phone number: ")
            Address=input("enter a address")
            status=input("Enter a Status : ")
            m = Manager( emp_id,name,salary,phone,Address,status, dept)
            employees.append(m)

        elif choice == "2":
            name = input("Enter Developer name: ")
            emp_id = input("Enter ID: ")
            salary = float(input("Enter salary: "))
            phone=input("Enter a Phone number: ")
            Address=input("enter a address")
            status=input("Enter a Status : ")
            lang = input("Enter programming language: ")
            d = Developer(emp_id,name,salary,phone,Address, status,lang)
            employees.append(d)

        elif choice == "3":
            emp_id = input("Enter employee ID to update salary: ")
            for e in employees:
                if e.emp_id == emp_id:
                    new_salary = float(input("Enter new salary: "))
                    e.setSalary(new_salary)
                    print("Salary updated")
                    break
            else:
                print("Employee not found")

        elif choice == "4":
            for e in employees:
                e.display()

        elif choice == "5":
            print("Exiting...")
            break

        else:
            print(" Invalid choice")


if __name__ == "__main__":
    main()

# Example usage
#m = Manager(101, "Ali", 80000, "0300-1234567", "Lahore","Manager" ,"HR")
#d = Developer(102, "Hashim", 90000, "0312-9876543", "Karachi", "Developer","Python")

#m.display()
#d.display()
```

```python
from abc import ABC, abstractmethod

# ---------------- Custom Exceptions ----------------
class InvalidInputError(Exception): pass
class InsufficientQuantityError(Exception): pass
class CustomerNotFoundError(Exception): pass
class EmployeeNotFoundError(Exception): pass
class ItemNotFoundError(Exception): pass

# ---------------- Abstract Base Class ----------------
class Person(ABC):
    @abstractmethod
    def display_info(self): pass

# ---------------- Employee Class ----------------
class Employee(Person):
    def __init__(self, emp_id, name, role, salary):
        self.emp_id = str(emp_id).strip()
        self.name = name.strip()
        self.role = role.strip()
        self._salary = float(salary)

    def display_info(self):
        print(self)

    def __str__(self):
        return f"ID: {self.emp_id} | Name: {self.name} | Role: {self.role} | Salary: ${self._salary:,.2f}"

    def get_salary(self):
        return self._salary

# ---------------- Inventory Item ----------------
class InventoryItem:
    def __init__(self, item_id, name, quantity, price):
        self.item_id = str(item_id).strip()
        self.name = name.strip()
        self.quantity = int(quantity)
        self.price = float(price)

    def display(self):
        print(self)

    def __str__(self):
        return f"ID: {self.item_id} | Name: {self.name} | Qty: {self.quantity} | Price: ${self.price:,.2f}"

# ---------------- Customer Class ----------------
class Customer(Person):
    def __init__(self, cust_id, name):
        self.cust_id = str(cust_id).strip()
        self.name = name.strip()
        self._purchases = {}

    def display_info(self):
        total_items = sum(self._purchases.values())
        print(f"ID: {self.cust_id} | Name: {self.name} | Total Items Purchased: {total_items}")

    def purchase(self, item, quantity):
        if quantity <= 0:
            raise InvalidInputError("Quantity must be positive.")
        if quantity > item.quantity:
            raise InsufficientQuantityError(f"Not enough {item.name} in stock. Available: {item.quantity}")
        
        item.quantity -= quantity
        self._purchases[item.item_id] = self._purchases.get(item.item_id, 0) + quantity
        total_cost = item.price * quantity
        print(f"{self.name} purchased {quantity} of {item.name} for ${total_cost:,.2f}")

    def __call__(self):
        print(f"\n{self.name}'s Purchase Summary:")
        if not self._purchases:
            print("No purchases yet.")
            return
        for item_id, qty in self._purchases.items():
            print(f"Item ID: {item_id} | Quantity: {qty}")

# ---------------- Management System ----------------
class ManagementSystem:
    def __init__(self):
        self.employees = {}
        self.inventory = {}
        self.customers = {}

    def add_employee(self, emp):
        if emp.emp_id in self.employees:
            raise InvalidInputError("Employee with this ID already exists!")
        self.employees[emp.emp_id] = emp
        print("Employee added successfully!")

    def display_employees(self):
        if not self.employees:
            print("No employees found!")
            return
        print("\n--- Employees ---")
        for emp in self.employees.values():
            emp.display_info()

    def highest_paid_employee(self):
        if not self.employees:
            raise EmployeeNotFoundError("No employees found!")
        highest = max(self.employees.values(), key=lambda e: e.get_salary())
        print("\n--- Highest Paid Employee ---")
        highest.display_info()
        return highest

    def add_inventory_item(self, item):
        if item.item_id in self.inventory:
            raise InvalidInputError("Item with this ID already exists!")
        self.inventory[item.item_id] = item
        print("Inventory item added successfully!")

    def display_inventory(self):
        if not self.inventory:
            print("No inventory items found!")
            return
        print("\n--- Inventory ---")
        for item in self.inventory.values():
            item.display()

    def get_low_stock_items(self, threshold=5):
        low_stock = [item for item in self.inventory.values() if item.quantity < threshold]
        if low_stock:
            print(f"\n--- Low Stock Items (below {threshold}) ---")
            for item in low_stock:
                item.display()
        return low_stock

    def add_customer(self, cust):
        if cust.cust_id in self.customers:
            raise InvalidInputError("Customer with this ID already exists!")
        self.customers[cust.cust_id] = cust
        print("Customer added successfully!")

    def display_customers(self):
        if not self.customers:
            print("No customers found!")
            return
        print("\n--- Customers ---")
        for cust in self.customers.values():
            cust.display_info()

    def dashboard(self):
        print("\n=== SYSTEM DASHBOARD ===")
        print(f"Total Employees: {len(self.employees)}")
        print(f"Total Inventory Items: {len(self.inventory)}")
        print(f"Total Customers: {len(self.customers)}")
        total_stock_value = sum(item.price * item.quantity for item in self.inventory.values())
        print(f"Total Stock Value: ${total_stock_value:,.2f}")

# ---------------- Interactive Console ----------------
def main():
    system = ManagementSystem()

    while True:
        print("\n" + "="*50)
        print("          MANAGEMENT SYSTEM MENU")
        print("="*50)
        print("1. Add Employee")
        print("2. Add Inventory Item")
        print("3. Add Customer")
        print("4. Display Employees")
        print("5. Display Inventory")
        print("6. Display Customers")
        print("7. Purchase Item")
        print("8. Highest Paid Employee")
        print("9. Dashboard")
        print("10. Low Stock Alert")
        print("11. Exit")
        print("="*50)

        choice = input("Enter your choice (1-11): ").strip()

        try:
            if choice == "1":
                emp_id = input("Employee ID: ").strip()
                name = input("Name: ").strip()
                role = input("Role: ").strip()
                salary = float(input("Salary: $"))
                system.add_employee(Employee(emp_id, name, role, salary))

            elif choice == "2":
                item_id = input("Item ID: ").strip()
                name = input("Item Name: ").strip()
                quantity = int(input("Quantity: "))
                price = float(input("Price: $"))
                system.add_inventory_item(InventoryItem(item_id, name, quantity, price))

            elif choice == "3":
                cust_id = input("Customer ID: ").strip()
                name = input("Customer Name: ").strip()
                system.add_customer(Customer(cust_id, name))

            elif choice == "4":
                system.display_employees()

            elif choice == "5":
                system.display_inventory()

            elif choice == "6":
                system.display_customers()

            elif choice == "7":
                if not system.customers:
                    print("No customers available!")
                    continue
                if not system.inventory:
                    print("No inventory items available!")
                    continue

                system.display_customers()
                cust_id = input("\nCustomer ID: ").strip()
                
                # FIXED: Use a more robust way to find the customer
                customer = None
                for cust_key, cust_obj in system.customers.items():
                    if cust_key == cust_id:
                        customer = cust_obj
                        break
                
                if not customer:
                    print(f"DEBUG: Available Customer IDs: {list(system.customers.keys())}")
                    print(f"DEBUG: Looking for: '{cust_id}'")
                    raise CustomerNotFoundError(f"Customer with ID '{cust_id}' not found!")

                system.display_inventory()
                item_id = input("\nItem ID to purchase: ").strip()
                
                # FIXED: Use a more robust way to find the item
                item = None
                for item_key, item_obj in system.inventory.items():
                    if item_key == item_id:
                        item = item_obj
                        break
                
                if not item:
                    print(f"DEBUG: Available Item IDs: {list(system.inventory.keys())}")
                    print(f"DEBUG: Looking for: '{item_id}'")
                    raise ItemNotFoundError(f"Item with ID '{item_id}' not found!")

                quantity = int(input("Quantity to purchase: "))
                customer.purchase(item, quantity)
                customer()

            elif choice == "8":
                system.highest_paid_employee()

            elif choice == "9":
                system.dashboard()

            elif choice == "10":
                system.get_low_stock_items()

            elif choice == "11":
                print("Thank you for using the Management System. Goodbye!")
                break

            else:
                raise InvalidInputError("Please choose a valid option (1-11)")

        except (ValueError, InvalidInputError, InsufficientQuantityError, 
                CustomerNotFoundError, EmployeeNotFoundError, ItemNotFoundError) as e:
            print(f" Error: {e}")
        except Exception as e:
            print(f" Unexpected Error: {e}")

if __name__ == "__main__":
    main()
```
