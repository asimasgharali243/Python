```python
import pandas as pd
df=pd.read_csv(r'C:\Users\World\Downloads\Compressed/retail_store_sales.csv')
df
```

```python
df.isna().mean()*100
```

```python
df.isna().sum().to_frame(name="null_count").assign(percentage_null=lambda x: (x["null_count"] / len(df) * 100))
```

```python
df['Item'].fillna(df['Item'].mode()[0],inplace=True)
df[['Transaction ID','Item','Price Per Unit']]
```

```python
df['Quantity'].fillna(df['Quantity'].median(),inplace=True)
df[['Transaction ID','Item','Quantity']]
```

```python
df['Total Spent'] = df['Total Spent'].fillna(df['Quantity'] * df['Price Per Unit'])
df['Total Spent'].fillna(0,inplace=True)
df['Total Spent'].to_frame(name='Total Spent')
```

```python
df['Discount Applied'].isna().sum()/len(df)*100
df['Discount Applied'].isna().mean()*100
df['Discount Applied'].fillna(False,inplace=True)
df[['Transaction ID','Item','Quantity','Discount Applied']]
```

```python
df.isna().sum()
```

```python
mask = df['Price Per Unit'].isna() & df['Quantity'].notna() & df['Total Spent'].notna()
df.loc[mask, 'Price Per Unit'] = df.loc[mask, 'Total Spent'] / df.loc[mask, 'Quantity']
df
```
