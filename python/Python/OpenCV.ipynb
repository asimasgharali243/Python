```python
# --- 1. Translation (move) ---
M = np.float32([[1, 0, 100], [0, 1, 50]])  # Right 100, Down 50
translated = cv2.warpAffine(img, M, (w, h))

# --- 2. Rotation ---
center = (w // 2, h // 2)  # rotate around center
angle = 45  # degrees
scale = 1.0
R = cv2.getRotationMatrix2D(center, angle, scale)
rotated = cv2.warpAffine(img, R, (w, h))

# --- 3. Scaling (resizing) ---
scaled_up = cv2.resize(img, None, fx=1.5, fy=1.5)   # 1.5x bigger
scaled_down = cv2.resize(img, None, fx=0.5, fy=0.5) # 0.5x smaller

# --- 4. Flipping ---
flip_horizontal = cv2.flip(img, 1)   # left-right
flip_vertical = cv2.flip(img, 0)     # up-down
flip_both = cv2.flip(img, -1)        # both axes

# Show results
cv2.imshow("Translated", translated)
cv2.imshow("Rotated", rotated)
cv2.imshow("Scaled Up", scaled_up)
cv2.imshow("Scaled Down", scaled_down)
cv2.imshow("Flip Horizontal", flip_horizontal)
cv2.imshow("Flip Vertical", flip_vertical)
cv2.imshow("Flip Both", flip_both)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import numpy as np
import cv2

def rotation_matrix_about_point(x0, y0, theta_deg, scale=1.0):
    # Convert angle to radians
    theta = np.deg2rad(theta_deg)

    # Step 1: Translation to origin
    T1 = np.array([
        [1, 0, -x0],
        [0, 1, -y0],
        [0, 0,  1]
    ])

    # Step 2: Rotation (and scaling)
    R = np.array([
        [scale*np.cos(theta), -scale*np.sin(theta), 0],
        [scale*np.sin(theta),  scale*np.cos(theta), 0],
        [0, 0, 1]
    ])

    # Step 3: Translate back
    T2 = np.array([
        [1, 0, x0],
        [0, 1, y0],
        [0, 0,  1]
    ])

    # Final Matrix: T2 * R * T1
    M = T2 @ R @ T1
    return M


w, h = 400, 300   # example image size
x0, y0 = w/2, h/2
theta = 45  # rotate 45°

M = rotation_matrix_about_point(x0, y0, theta)

print("3x3 Homogeneous Matrix:\n", M)

# Convert to 2x3 for OpenCV warpAffine
M_affine = M[:2,:]
print("\nEquivalent 2x3 Affine Matrix:\n", M_affine)

# Compare with OpenCV built-in function
M_cv2 = cv2.getRotationMatrix2D(center=(x0, y0), angle=theta, scale=1.0)
print("\nOpenCV getRotationMatrix2D:\n", M_cv2)

```
