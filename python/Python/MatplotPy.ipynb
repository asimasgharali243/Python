```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Reproducibility
np.random.seed(42)

# --- Step 1: Create Base Data ---
n = 10000
income = np.random.normal(50000, 8000, n).round(2)
days_on_platform = np.random.randint(30, 365, n)
age = np.random.randint(18, 60, n).astype(float)
visits = np.random.randint(5, 50, n).astype(float)

# Purchases strongly correlated with income and days_on_platform
purchases = (0.00025 * income + 0.04 * days_on_platform + np.random.normal(0, 2, n)).round(0)
purchases = np.where(purchases < 0, 0, purchases)

# Column that needs log transformation (right-skewed)
spending_score = np.random.exponential(scale=100, size=n).round(2)

# --- Step 2: Introduce Outliers ---
income[np.random.choice(n, 5, replace=False)] *= 5   # Extreme high incomes
visits[np.random.choice(n, 5, replace=False)] *= 10  # Extreme visit counts

# --- Step 3: Introduce Missing Values ---
for col in ["income", "age", "visits"]:
    missing_idx = np.random.choice(n, size=20, replace=False)
    if col == "income":
        income[missing_idx] = np.nan
    elif col == "age":
        age[missing_idx] = np.nan
    elif col == "visits":
        visits[missing_idx] = np.nan

# --- Step 4: Combine into DataFrame ---
synthetic_df = pd.DataFrame({
    "id": np.arange(1, n+1),
    "age": age,
    "income": income,
    "days_on_platform": days_on_platform,
    "visits": visits,
    "spending_score": spending_score,
    "purchases": purchases
})

# --- Step 5: Save to CSV ---
file_path = r"C:\Users\World\Downloads\synthetic_clv_training_data.csv"
synthetic_df.to_csv(file_path, index=False)

print(f"✅ Synthetic dataset created and saved to: {file_path}")
print(synthetic_df.head())
print("\nMissing values per column:\n", synthetic_df.isnull().sum())

# --- Step 6: Plot Histogram ---
data = np.random.randn(1000)  # random normal distribution
plt.hist(data, bins=20, edgecolor='black')
plt.title("Histogram - Distribution of Data")
plt.show()
```
