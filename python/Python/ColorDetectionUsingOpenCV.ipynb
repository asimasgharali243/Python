```python
import cv2
import numpy as np
import pytesseract
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"
import re


# Initialize webcam
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Resize for faster processing
    frame = cv2.resize(frame, (640, 480))

    # ----------- 1️⃣ COLOR DETECTION -----------
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # Define LED color ranges (adjust if needed)
    colors = {
        "Red": [(0, 100, 100), (10, 255, 255)],
        "Green": [(36, 100, 100), (86, 255, 255)],
        "Blue": [(94, 80, 2), (126, 255, 255)],
        "Yellow": [(15, 100, 100), (35, 255, 255)]
    }

    detected_color = None
    for color, (lower, upper) in colors.items():
        mask = cv2.inRange(hsv, np.array(lower), np.array(upper))
        count = cv2.countNonZero(mask)
        if count > 500:  # threshold to ignore small noise
            detected_color = color
            cv2.putText(frame, f"LED Color: {color}", (20, 420),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 255), 2)
            break

    # ----------- 2️⃣ WEIGHT DETECTION -----------
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    blur = cv2.GaussianBlur(gray, (5, 5), 0)
    _, thresh = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

    config = "--psm 7 -c tessedit_char_whitelist=0123456789."
    text = pytesseract.image_to_string(thresh, config=config)

    # Extract numeric value using regex
    match = re.findall(r"\d+\.\d+|\d+", text)
    if match:
        weight = match[0]
        cv2.putText(frame, f"Weight: {weight} kg", (20, 50),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
        print(f"Detected Weight: {weight} | LED Color: {detected_color}")
    else:
        cv2.putText(frame, "Weight: Not detected", (20, 50),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)

    # ----------- 3️⃣ DISPLAY WINDOWS -----------
    cv2.imshow("Processed View", thresh)
    cv2.imshow("Live Feed", frame)

    if cv2.waitKey(1) & 0xFF == 27:  # ESC to quit
        break

cap.release()
cv2.destroyAllWindows()
```
