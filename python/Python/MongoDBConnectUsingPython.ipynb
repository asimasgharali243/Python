```python
import speech_recognition as sr
import pyttsx3
from difflib import SequenceMatcher
import pyaudio
import wave

# ------------------------------
# Initialize Text-to-Speech (TTS)
# ------------------------------
engine = pyttsx3.init()
engine.setProperty('rate', 150)

def speak(text):
    """Speak out the given text."""
    engine.say(text)
    engine.runAndWait()

# ------------------------------
# Initialize Speech Recognizer
# ------------------------------
recognizer = sr.Recognizer()

def list_microphones():
    """List all available microphone devices."""
    print("Available Microphones:")
    for i, mic_name in enumerate(sr.Microphone.list_microphone_names()):
        print(f"{i}: {mic_name}")

def choose_microphone():
    """Ask user to choose a microphone index."""
    list_microphones()
    while True:
        try:
            index = int(input("Enter the microphone index you want to use: "))
            return index
        except ValueError:
            print("Please enter a valid number.")

# ------------------------------
# Audio Quality Check
# ------------------------------
def check_microphone_quality(device_index=None):
    """Check if microphone is working properly."""
    try:
        p = pyaudio.PyAudio()
        stream = p.open(format=pyaudio.paInt16,
                       channels=1,
                       rate=44100,
                       input=True,
                       input_device_index=device_index,
                       frames_per_buffer=1024)
        
        print("Microphone test - recording 2 seconds...")
        frames = []
        for _ in range(0, int(44100 / 1024 * 2)):
            data = stream.read(1024)
            frames.append(data)
        
        stream.stop_stream()
        stream.close()
        p.terminate()
        
        # Save test recording
        with wave.open("test_mic.wav", 'wb') as wf:
            wf.setnchannels(1)
            wf.setsampwidth(p.get_sample_size(pyaudio.paInt16))
            wf.setframerate(44100)
            wf.writeframes(b''.join(frames))
        
        print("Microphone test completed. Check test_mic.wav file.")
        return True
    except Exception as e:
        print(f"Microphone test failed: {e}")
        return False

# ------------------------------
# Enhanced Listening Function
# ------------------------------
def enhanced_listen(device_index=None, max_attempts=3):
    """Enhanced listening with multiple attempts and fallbacks."""
    
    for attempt in range(max_attempts):
        with sr.Microphone(device_index=device_index) as source:
            print(f"Attempt {attempt + 1}/{max_attempts}")
            recognizer.adjust_for_ambient_noise(source, duration=1.5)
            
            # Adjust settings for better recognition
            recognizer.energy_threshold = 4000
            recognizer.pause_threshold = 0.8
            
            print("Listening... (speak clearly)")
            try:
                audio = recognizer.listen(source, timeout=15, phrase_time_limit=6)
            except sr.WaitTimeoutError:
                print("No speech detected. Try again.")
                continue
        
        try:
            # Primary recognition with Google
            text = recognizer.recognize_google(audio)
            print(f"You said: {text}")
            return text
        except sr.UnknownValueError:
            if attempt < max_attempts - 1:
                print("Could not understand. Please try again.")
                speak("I didn't catch that. Please repeat.")
            else:
                print("Maximum attempts reached. Moving on.")
        except sr.RequestError as e:
            print(f"Service error: {e}")
            break
    
    return ""

# ------------------------------
# Pronunciation Scoring
# ------------------------------
def pronunciation_score(user_text, correct_text):
    """Compare user text with correct text and return similarity ratio."""
    if not user_text:  # Handle empty input
        return 0.0
    return SequenceMatcher(None, user_text.lower(), correct_text.lower()).ratio()

# ------------------------------
# Improved Lesson Function
# ------------------------------
def run_improved_lesson(lesson_word, mic_index=None):
    """Run lesson with better error handling and user feedback."""
    
    # Check microphone first (skip if pyaudio not available)
    try:
        if not check_microphone_quality(mic_index):
            print("Microphone issues detected. Please check your microphone.")
            return
    except NameError:
        print("Skipping microphone quality check (pyaudio not available)")
    
    speak(f"Please say the word: {lesson_word}")
    print(f"Expected word: {lesson_word}")
    
    user_input = enhanced_listen(device_index=mic_index)
    
    if user_input:
        score = pronunciation_score(user_input, lesson_word)
        print(f"Pronunciation score: {score:.2f}")
        
        if score > 0.8:
            speak("Excellent pronunciation! Well done!")
        elif score > 0.6:
            speak("Good effort! Let's practice some more.")
            speak(f"You said: {user_input}. Try to match: {lesson_word}")
        else:
            speak("Let's try again. Listen carefully.")
            speak(f"The word is: {lesson_word}")
            speak("Now you try...")
            
            # Second attempt
            user_input_2 = enhanced_listen(device_index=mic_index)
            if user_input_2:
                score_2 = pronunciation_score(user_input_2, lesson_word)
                if score_2 > score:
                    speak("Much better! You're improving.")
                else:
                    speak("Keep practicing. You'll get it!")
    else:
        speak("I couldn't hear anything. Let's try another time.")

# ------------------------------
# Simple version without pyaudio dependency
# ------------------------------
def run_simple_lesson(lesson_word, mic_index=None):
    """Run lesson without microphone quality check."""
    
    speak(f"Please say the word: {lesson_word}")
    print(f"Expected word: {lesson_word}")
    
    # Use the original listen function or enhanced_listen
    user_input = enhanced_listen(device_index=mic_index)
    
    if user_input:
        score = pronunciation_score(user_input, lesson_word)
        print(f"Pronunciation score: {score:.2f}")
        
        if score > 0.8:
            speak("Excellent pronunciation! Well done!")
        elif score > 0.6:
            speak("Good effort! Let's practice some more.")
            speak(f"You said: {user_input}. Try to match: {lesson_word}")
        else:
            speak("Let's try again. Listen carefully.")
            speak(f"The word is: {lesson_word}")
            speak("Now you try...")
            
            # Second attempt
            user_input_2 = enhanced_listen(device_index=mic_index)
            if user_input_2:
                score_2 = pronunciation_score(user_input_2, lesson_word)
                if score_2 > score:
                    speak("Much better! You're improving.")
                else:
                    speak("Keep practicing. You'll get it!")
    else:
        speak("I couldn't hear anything. Let's try another time.")

# ------------------------------
# Program Entry Point
# ------------------------------
if __name__ == "__main__":
    print("=== Improved Pronunciation Lesson ===")
    
    try:
        mic_index = choose_microphone()
        lesson_words = ["Hello", "World", "Python", "Program"]
        
        for word in lesson_words:
            print(f"\n--- Practicing: {word} ---")
            # Use simple version to avoid pyaudio dependency issues
            run_simple_lesson(word, mic_index)
            
    except KeyboardInterrupt:
        print("\nLesson ended by user.")
    except Exception as e:
        print(f"Unexpected error: {e}")
```

```python
import speech_recognition as sr

def test_microphone(device_index=None):
    recognizer = sr.Recognizer()

    try:
        # List microphones if index not provided
        if device_index is None:
            print("Available microphones:")
            for i, name in enumerate(sr.Microphone.list_microphone_names()):
                print(f"{i}: {name}")
            device_index = int(input("Enter microphone index to test: "))

        # Access the microphone
        with sr.Microphone(device_index=device_index) as source:
            recognizer.adjust_for_ambient_noise(source, duration=1)
            print("Microphone is ready. Speak something (5 seconds)...")
            audio = recognizer.listen(source, timeout=5, phrase_time_limit=5)

        # Try converting speech to text
        text = recognizer.recognize_google(audio)
        print(f"✅ Success! You said: {text}")

    except sr.WaitTimeoutError:
        print("❌ No speech detected. Make sure you start speaking after prompt.")
    except sr.UnknownValueError:
        print("❌ Could not understand audio. Try speaking more clearly or closer to the mic.")
    except sr.RequestError:
        print("❌ Could not contact Google Speech Recognition service. Check your internet connection.")
    except Exception as e:
        print(f"❌ Microphone test failed: {e}")
        print("Check that your microphone is connected, enabled, and not being used by another app.")

if __name__ == "__main__":
    test_microphone()

```
