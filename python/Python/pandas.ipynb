```python
import pandas as pd

df2 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Class': [11, 12, 13, 14]
})

df3 = pd.DataFrame({
    'ID': [5, 6, 7, 8],
    'Class': [11, 12, 13, 14]
})
print(df2)
print(df3)

merged_default = pd.merge(df2, df3)
merged_index = pd.merge(df2, df3, left_index=True, right_index=True)

print("Default merge:")
print(merged_default)

print("\nMerge using index:")
print(merged_index)
```

```python

df3 = pd.DataFrame({ 'ID':[5,6,7,8],'Class':[11,12,13,14]})
df4 = pd.DataFrame({ 'ID':[5,6,7,8],'Class':[11,12,13,14]})

merged_default = pd.merge(df3, df4, on='ID')
print(merged_default)
```

```python

sr1 = pd.Series([0,1,2])
sr2 = pd.Series([3,4,5])

pd.concat([sr1, sr2])
```

```python
pd.concat([sr1, sr2], ignore_index=True)
```

```python
df5 = pd.DataFrame({'ID':[1,2,3,4],'Name':['A','B','C','D'],'Class':['w','x','y','z']})
df6 = pd.DataFrame({'ID':[5,6,7,8],'Name':['D','E','F','G'],'Class':['W','X','Y','Z']})

pd.concat([df6, df5])
```

```python
pd.concat([df6, df5], axis=1)
```

```python
df7 = pd.DataFrame({'ID':[1,2,3,4],'Name':['A','B','C','D'],'Class':['w','x','y','z']})
df8 = pd.DataFrame({'ID':[3,4],'Name':['C','D'],'Class':['y','z']})

pd.concat([df7, df8], axis=1, join='outer', keys=['Student','Teacher'])
```

```python
df10 = pd.DataFrame({'A':[1,2,3,4],'B':[11,12,13,14]}, index=['a','b','c','d'])
df11 = pd.DataFrame({'C':[5,6],'D':[11,12]}, index=['a','b'])

print(df10.join(df11))
```

```python

df10.join(df11, lsuffix='1')
df10.join(df11, rsuffix='_1')
```
