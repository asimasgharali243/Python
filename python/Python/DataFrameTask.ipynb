```python
summary_df = (
    pf1.groupby('Category')
       .agg(
           Total_Products=('Name', 'count'),
           Avg_Price=('Price', 'mean'),
           Total_Stock=('Stock', 'sum'),
           Min_Price=('Price', 'min'),
           Max_Price=('Price', 'max')
       )
       .reset_index()
)
summary_df 
```

```python
availability_counts = (
    pf1.groupby(['Category', 'Availability'])
        .size()
        .unstack(fill_value=0)  
)
availability_counts['InStock_to_PreOrder_Ratio'] = (
    availability_counts['in_stock'] / availability_counts['pre_order']
)
availability_counts[['in_stock', 'pre_order', 'InStock_to_PreOrder_Ratio']]
```
