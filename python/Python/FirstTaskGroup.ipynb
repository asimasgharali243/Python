```python
# task 6
#variable
marks = {'english': '', 'math': '', 'islamiate': '', 'physics': '', 'computer': '', 'urdu': ''}

for subject in marks:
    marks[subject] = float(input(f"Enter {subject} marks: "))

#logic
if (marks['english'] < 40 or marks['math'] < 40 or marks['islamiate'] < 40 
    or marks['physics'] < 40 or marks['computer'] < 40 or marks['urdu'] < 40):
    
    print("You are Failed")

else:
    if (marks['english'] > 60 and marks['math'] > 60 and marks['islamiate'] > 60
        and marks['physics'] > 60 and marks['computer'] > 60 and marks['urdu'] > 60):
        
        print("You are First Division")

    else:
        if (marks['english'] > 50 and marks['math'] > 50 and marks['islamiate'] > 50
            and marks['physics'] > 50 and marks['computer'] > 50 and marks['urdu'] > 50):    
            
            print("You are Second Division")

        else:
            print("You are Third Division")

    
```

```python
#Task 7
#variable
temperature = float(input("Enter the temperature in °C: "))
weather = input("Enter the weather condition (sunny/rainy): ")

#logic
if temperature < 10:
    print("Advice: Wear heavy jackets.")
else:
    if temperature >= 10:
        if temperature <= 25:
            if weather == "rainy":
                print("Advice: Carry an umbrella and wear a light sweater.")
            else:
                print("Advice: Wear a light sweater.")
        else:
            if temperature > 25:
                if temperature <= 35:
                    if weather == "sunny":
                        print("Advice: Good time for an outing!")
                    else:
                        print("Advice: Weather not suitable for outing.")
                else:
                    if temperature > 35:
                        print("Advice: Stay indoors due to heat.")
```

```python
#Task 11
#variable
bill = float(input("Enter total bill amount: "))
payment = input("Payment method (cash/credit): ").lower()
#logic
if bill >= 5000:
    discount = bill * 0.20
    bill = bill - discount
else:
    if bill >= 3000:
        discount = bill * 0.10
        bill = bill - discount
    else:
        discount = 0  

if payment == "credit":
    extra_discount = bill * 0.05
    bill = bill - extra_discount
else:
    extra_discount = 0

print("Final Bill after discounts:", bill)
```

```python
#Task 14
#variable
room_type = (input("Enter room type (Luxury/Deluxe/Standard): ")).lower()
nights = int(input("Enter number of nights: "))

if room_type == "luxury":
    cost_per_night = 10000
else:
    if room_type == "deluxe":
        cost_per_night = 6000
    else:
        if room_type == "standard":
            cost_per_night = 3000
        else:
            print("Invalid room type")
            cost_per_night = 0

if cost_per_night > 0:
    total_cost = cost_per_night * nights

    if nights > 5:
        discount = total_cost * 0.15
        total_cost = total_cost - discount
    else:
        total_cost = total_cost  

    print("Total Room Charges: Rs.", total_cost)
else:
    print("Booking failed due to invalid room type")
```
