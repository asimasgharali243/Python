```python
import pandas as pd
import re
df=pd.read_csv(r'C:\Users\World\Downloads\Compressed/audible_uncleaned.csv')
df
```

```python
df['time'] = df['time'].astype(str)

h = df['time'].str.extract(r'(\d+)\s*hrs?')[0].fillna(0).astype(int)
m = df['time'].str.extract(r'(\d+)\s*mins?')[0].fillna(0).astype(int)

df['time'] = pd.to_timedelta(h * 60 + m, unit='m')

df['time']
```

```python
df['releasedate'] = pd.to_datetime(df['releasedate'], format='%d-%m-%y', errors='coerce')
df
```

```python
import numpy as np

df['stars'] = df['stars'].astype(str)

df['stars'] = df['stars'].replace(r'(?i)^not rated yet$', np.nan, regex=True)

df['stars'] = df['stars'].str.replace(
    r'^.*?(\d+\.?\d*)\s*out of\s*5.*$',  
    r'\1',
    regex=True
)

df['stars'] = df['stars'].replace('<NA>', np.nan)

df['stars'] = df['stars'].astype(float)

df
```

```python
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df
```

```python
df['language'] = df['language'].astype('category')
df
```

```python
empty_rows = df[df.isna().all(axis=1)]
print("Completely empty rows:", empty_rows)
```

```python
empty_columns = df.columns[df.isna().all(axis=0)]
print("Completely empty columns:", empty_columns.tolist())
```

```python
df
```

```python
df = df.dropna(how='all').dropna(axis=1, how='all')
df
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
df = df.rename(columns={'releasedate': 'release_date', 'stars': 'rating_text'})
df
```

```python
df["author"] = df["author"].str.replace(r"^Writtenby:", "", regex=True)
df["narrator"] = df["narrator"].str.replace(r"^Narratedby:", "", regex=True)
df
```

```python
df[["title", "series_info"]] = df["name"].str.extract(
    r'^(.*?)(?:\s*(?:[:,-]\s*|\s+#|,\s*Book\s*)(.*))?$'
)
df
```

```python
df["duration_minutes"] = (df["time"].dt.total_seconds() / 60).astype(int)
df
```

```python
print(df["release_date"].head(10))
mask = df["release_date"].astype(str).str.match(r"^\d{4}-\d{2}-\d{2}$")
print("Consistent ISO format:", mask.all())
```

```python
df["release_date"] = pd.to_datetime(df["release_date"], errors="coerce", infer_datetime_format=True)
df
```

```python
df["release_year"] = df["release_date"].dt.year
df["release_month"] = df["release_date"].dt.month
df["release_day"] = df["release_date"].dt.day
df["release_weekday"] = df["release_date"].dt.day_name()
df["release_quarter"] = df["release_date"].dt.quarter
df
```

```python
invalid_dates = df[df["release_date"].isna()]
print(invalid_dates)
```

```python
if (df['release_month'] > 12).any() or (df['release_month'] < 1).any():
    print("Wrong format: some months are invalid")
else:
    print("Correct format: all months are valid")
```

```python
missing_indicators = ["", "NA", "N/A", "None", "none", "-"]
df.replace(missing_indicators, np.nan, inplace=True)
df
```

```python
missing_counts = df.isna().sum()
missing_percent = (df.isna().mean()* 100)
missing_summary = pd.DataFrame({"missing_count": missing_counts, "missing_percent": missing_percent})
print(missing_summary)
```

```python
df['rating_text'] = df['rating_text'].fillna( df['rating_text'].median())
df['series_info']=df['series_info'].fillna('unknown')
df
```

```python
df['price'] = df['price'].fillna(df['price'].median())
df
```

```python
df['language'] = df['language'].str.title()
df['author'] = df['author'].str.title().str
df['narrator'] = df['narrator'].str.title()
df
```

```python
df["author"] = df["author"].str.replace(r'([a-z])([A-Z])', r'\1 \2', regex=True)
df["narrator"] = df["narrator"].str.replace(r'([a-z])([A-Z])', r'\1 \2', regex=True)
df['author'] = df['author'].str.replace(r'\.', '. ', regex=True).str.replace(r'\s+', ' ', regex=True).str.strip()
df['narrator'] = df['narrator'].str.replace(r'\.', '. ', regex=True).str.replace(r'\s+', ' ', regex=True).str.strip()

df["series_info"] = df["series_info"].str.replace(r'\s*([&#])\s*', r' \1 ', regex=True)

html_entities = r'&(?:amp|nbsp|lt|gt|quot|apos|copy|reg);'
df = df.apply(lambda col: col.str.replace(html_entities, ' ', regex=True) if col.dtype == 'object' else col)

df = df.apply(lambda col: col.str.strip().str.replace(r'\s+', ' ', regex=True) if col.dtype == 'object' else col)

df["title"] = df["title"].str.replace(r'^[:\s]+|[:\s]+$', '', regex=True)
df
```

```python
df['price'] = df['price'].astype(str).str.replace(r'[^\d.]','', regex=True).astype(float)
df
```
