```python
#. Minimum of three numbers
minimum = lambda a, b, c: a if a < b and a < c else (b if b < c else c)
print(minimum(5, 2, 8))   
```

```python
is_even = lambda x: "Even" if x % 2 == 0 else "Odd"
print(is_even(7))   
print(is_even(10))  
```

```python
#square of all 
numbers = [1, 2, 3, 4, 5]

squared = map(lambda x: x**2, numbers)

print(list(squared))
```

```python
#lambda with ,map and add all list 
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list3=[5,6,2]
result = map(lambda x, y,z: x + y+z, list1, list2,list3)

print(list(result))
```

```python
numbers = [1, 2,2,2, 3, 4, 5, 6]

evens = filter(lambda x: x % 2 == 0, numbers)

print(set(evens))
```

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5,6,9,10]

result = reduce(lambda x, y: x + y, numbers)

print(result)
```

```python
from functools import reduce

numbers = [7, 2, 9, 4, 5]

maximum = reduce(lambda a, b: a if a > b else b, numbers)

print(maximum)
```
