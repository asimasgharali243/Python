```python
Rdata["Stock"] = Rdata["Stock"].replace({
    "In Stock": 1,
    "Out of Stock": 0
}).infer_objects(copy=False)
```

```python
Rdata.fillna({
    "Category": "Unknown",             
    "Price": Rdata["Price"].mean(),  
    "Discount": Rdata["Discount"].median(),
    "Rating":Rdata['Rating'].mean(),
    "Stock":0,
    "Payment_Method": "Cash",         
}, inplace=True)
```

```python
import pandas as pd

df1 = pd.DataFrame({"A":[1,2], "B":[3,4]})
df2 = pd.DataFrame({"A":[5,6], "B":[7,8]})

result = pd.concat([df1, df2])
print(result)
```

```python
result = pd.concat([df1, df2], axis=1)
print(result)
```

```python
df3 = pd.DataFrame({"A":[9,10], "C":[11,12]})
result = pd.concat([df1, df3], join="inner", ignore_index=True)
print(result)
```
