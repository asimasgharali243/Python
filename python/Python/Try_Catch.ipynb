```python
import logging

# Configure logging
logging.basicConfig(
    filename="bank.log",
    level=logging.ERROR,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

# Custom Exception
class InsufficientFundsError(Exception):
    pass

def withdraw(balance, amount):
    try:
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive")
        if amount > balance:
            raise InsufficientFundsError("Not enough balance in account")
        balance -= amount
    except ValueError as ve:
        logging.error(f"Invalid withdrawal: {ve}")
        print("Error:", ve)
    except InsufficientFundsError as ife:
        logging.error(f"Insufficient funds: {ife}")
        print("Error:", ife)
    except Exception as e:
        logging.error(f"Unexpected error: {e}")
        print("An unexpected error occurred:", e)
    else:
        print(f"Withdrawal successful! New balance: {balance}")
    finally:
        print("Transaction attempt completed.")
    return balance


# 🔹 Usage
account_balance = 1000

# Valid withdrawal
account_balance = withdraw(account_balance, 200)

# Invalid (negative withdrawal)
account_balance = withdraw(account_balance, -50)

# Insufficient funds
account_balance = withdraw(account_balance, 5000)
```

```python
class NegativeNumberError(Exception):
    """Raised when a negative number is not allowed"""
    pass

def factorial(n):
    if n < 0:
        raise NegativeNumberError("Factorial is not defined for negative numbers")
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

try:
    print(factorial(-5))
except NegativeNumberError as e:
    print("❌ Custom Exception:", e)
```

```python
class Library:
    def __init__(self, name):
        self.name = name
        self.books = []   # list to store books

    def add_book(self, book):
        """Add a new book to the library"""
        self.books.append(book)
        print(f'Book "{book}" added to {self.name} library.')

    def display_books(self):
        """Show all available books"""
        if not self.books:
            print("No books available.")
        else:
            print(f"\nBooks in {self.name} library:")
            for idx, book in enumerate(self.books, 1):
                print(f"{idx}. {book}")

    def borrow_book(self, book):
        """Borrow a book if available"""
        if book in self.books:
            self.books.remove(book)
            print(f'You have borrowed "{book}".')
        else:
            print(f'Sorry, "{book}" is not available.')

    def return_book(self, book):
        """Return a borrowed book"""
        self.books.append(book)
        print(f'Thank you for returning "{book}".')


# ---------- Example Usage ----------
my_library = Library("Central")

# Add books
my_library.add_book("Python Programming")
my_library.add_book("Data Structures")
my_library.add_book("Machine Learning")

# Display
my_library.display_books()

# Borrow a book
my_library.borrow_book("Data Structures")

# After borrowing
my_library.display_books()

# Return a book
my_library.return_book("Data Structures")

# Final state
my_library.display_books()
```

```python
class Student:
    def __init__(self, name, rollNo, address):
        self.name = name
        self.rollNo = rollNo
        self.address = address
        self.marks = {}   
        self.grade = ""

    def add_marks(self, subject, score):
        """Add or update marks for a subject"""
        self.marks[subject] = score

    def calculate_grade(self):
        """Calculate grade based on average marks"""
        if not self.marks:
            self.grade = "No Marks"
            return self.grade

        total = sum(self.marks.values())
        avg = total / len(self.marks)

        if avg >= 90:
            self.grade = "A"
        elif avg >= 75:
            self.grade = "B"
        elif avg >= 60:
            self.grade = "C"
        elif avg >= 40:
            self.grade = "D"
        else:
            self.grade = "F"
        return self.grade

    def display(self):
        """Display student details"""
        print(f"\n--- Student Details ---")
        print(f"Name    : {self.name}")
        print(f"Roll No : {self.rollNo}")
        print(f"Address : {self.address}")
        print(f"Marks   : {self.marks}")
        print(f"Grade   : {self.grade}")


s1 = Student("Asim Ali", "BSCS-01", "Lahore")

s1.add_marks("Python", 85)
s1.add_marks("Math", 92)
s1.add_marks("Database", 78)
s1.add_marks("English", 78)
s1.add_marks("Urdu", 75)


s1.calculate_grade()

s1.display()
```

```python
class Account:
    # ---------- Constructor ----------
    def __init__(self, customer, balance):
        self.customer = customer       # Public variable
        self._balance = balance        # Protected variable
        self.__pin = "1234"            # Private variable
        print("Constructor called: Account Created")

    # ---------- Public Method ----------
    def deposit(self, amount):
        self._balance += amount
        print(f"{amount} deposited. Balance: {self._balance}")

    # ---------- Protected Method ----------
    def _show_balance(self):
        print(f"Balance: {self._balance}")

    # ---------- Private Method ----------
    def __show_pin(self):
        print(f"PIN: {self.__pin}")

    # ---------- Destructor ----------
    def __del__(self):
        print("Destructor called: Account Deleted")

# ----------- Usage -----------
acc = Account("Asim Ali", 5000)

# Public access
print(acc.customer)       # ✅ Works
acc.deposit(1000)         # ✅ Works

# Protected access (possible but discouraged)
print(acc._balance)       # ⚠️ Works but should not be used outside
acc._show_balance()

# Private access (Error if direct)
# print(acc.__pin)        # ❌ AttributeError

# Access private using name-mangling
print(acc._Account__pin)  # ✅ Works but not recommended
acc._Account__show_pin()  # ✅ Works but not recommended
```
